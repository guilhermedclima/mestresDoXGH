# 0. Criando o arquivo da aula

Rode esta célula primeiro. Ela gera a planilha `vendas.xlsx` que a gente vai usar.

````python
import pandas as pd, random

random.seed(11)
lojas = ["Centro", "Shopping", "Online"]

linhas = [
    {"mes": f"2026-{m:02d}",
     "loja": random.choice(lojas),
     "pecas": random.randint(10, 90),
     "faturamento": round(random.uniform(3000, 28000), 2)}
    for m in range(1, 13) for _ in range(3)
]

pd.DataFrame(linhas).to_excel("vendas.xlsx", index=False)
print("vendas.xlsx criado — 36 linhas, 4 colunas")
````

# 3. `pip install`: abrindo um formato novo

Sua planilha era `.csv` ou `.xlsx`? Quase sempre `.xlsx`.

A célula abaixo **vai dar erro de propósito** se o `openpyxl` não estiver instalado. Leia a mensagem:

````python
import pandas as pd

df = pd.read_excel("vendas.xlsx")
print(df.head(3))
````

```
ImportError: Missing optional dependency 'openpyxl'.
Use pip or conda to install openpyxl.
```

**Que erro educado.** Ele diz o que falta e o que fazer. Nem todo erro é assim — quando é, obedeça.

### Instalando

No terminal: `pip install openpyxl`

No Colab ou direto numa célula, use o `!` na frente:

````python
%pip install openpyxl
````

### ⚠️ `%pip` com porcentagem, não `!pip` com exclamação

Parece detalhe, mas é o 💥 nº 3 desta aula esperando por você:

| | O que faz |
|---|---|
| `%pip install` | instala **no kernel que está rodando este notebook** |
| `!pip install` | chama o pip que estiver no PATH — **pode ser de outro Python** |

Com `!pip` acontece o clássico: ele diz *Successfully installed*, você roda o `import`
e recebe `ModuleNotFoundError` mesmo assim. Instalou — só que no Python errado.

**No notebook, sempre `%pip`. No terminal, sempre `python -m pip`.**
Os dois garantem a mesma coisa: o pip e o código são do mesmo Python.

> Depois de instalar, **reinicie o kernel**. O Python só procura a biblioteca uma vez por sessão.

Agora rode de novo a célula do `read_excel` lá em cima. Deve funcionar.

````python
df = pd.read_excel("vendas.xlsx")
print(df.head(3))
print()
print(df.groupby("loja")["faturamento"].sum().round(2))
````

Um comando e o pandas passou a ler Excel.

Você **não aprendeu nada novo de pandas** — o `groupby` é o mesmo de sábado passado.
Só abriu um formato a mais.

> É isso que biblioteca faz: não te ensina a programar melhor, **te dá alcance**.

# 4. 💥 Os quatro jeitos de quebrar

`pip install` funciona em cinco segundos. O que toma tempo na vida real são estes quatro.
Você vai bater em todos — melhor bater aqui.

## 💥 nº 1 — a biblioteca que não existe

````python
import tabelinha_magica
````

```
ModuleNotFoundError: No module named 'tabelinha_magica'
```

Esse é o segundo erro mais comum do Python. Quer dizer **uma** de três coisas,
nesta ordem de probabilidade:

1. você não instalou
2. escreveu o nome errado
3. instalou no lugar errado

## 💥 nº 2 — nome de instalação ≠ nome de import

**Este é o que mais confunde.** Guarde esta tabela:

| Você instala | Você importa |
|---|---|
| `pip install scikit-learn` | `import sklearn` |
| `pip install pillow` | `import PIL` |
| `pip install beautifulsoup4` | `import bs4` |
| `pip install opencv-python` | `import cv2` |
| `pip install pyyaml` | `import yaml` |
| `pip install python-dateutil` | `import dateutil` |

**Não tem regra. Não dá para deduzir.** São nomes históricos.

E é exatamente por isso que você **lê a documentação**: a primeira linha de qualquer página de
biblioteca mostra o `pip install` e o `import`. Decorar é impossível; consultar leva 5 segundos.

## 💥 nº 3 — instalei e continua dizendo que não existe

O clássico frustrante: você instala, dá certo, roda o programa e ele jura que não existe.

**A causa:** o `pip` instalou num Python e o programa rodou em outro.

Como conferir:

````python
import sys
print("o Python que está rodando este código:")
print(" ", sys.executable)
````

E no terminal, `pip -V` mostra de qual Python é o pip.
**Se os dois caminhos não baterem, achou o problema.**

No VS Code: `Ctrl+Shift+P` → `Python: Select Interpreter`.
No Colab isso não acontece, porque só existe um Python lá.

### 🔑 A receita que resolve 90% dos casos

```
python -m pip install nome-da-biblioteca
```

`python -m pip` garante que o pip usado é o do **mesmo** Python que vai rodar seu código.
Pegue o hábito e você não vê mais esse erro.

## 💥 nº 4 — o pacote com nome parecido

E se eu digitar `pip install pandsa`, com as letras trocadas?

Às vezes: erro de não encontrado. Às vezes: **instala**.

Existe gente que registra pacote com nome quase igual ao de bibliotecas famosas, esperando o erro
de digitação. Chama **typosquatting**.

`pip install` roda com a sua máquina e as suas permissões. Então:

- use o link da **documentação oficial ou do PyPI**, não o primeiro resultado de busca
- olhe **downloads** e **data da última versão** na página do PyPI
- desconfie de pacote novo, com poucos downloads e nome parecido com um famoso

# 5. `requirements.txt` — o fim do "na minha máquina funciona"

Você termina o Projeto 2, manda para um colega, e ele abre: `ModuleNotFoundError`.
Óbvio — as bibliotecas estão na **sua** máquina.

Como você diz a ele o que instalar? Com uma lista:

```
pip freeze > requirements.txt
```

Gera um arquivo assim:

```
openpyxl==3.1.5
pandas==3.0.2
qrcode==8.2
rich==15.0.0
```

E do outro lado, um comando só:

```
pip install -r requirements.txt
```

### Por que a versão está ali

````python
import pandas as pd
print(pd.__version__)
````

Biblioteca muda. O que funcionava na versão 1 pode ter mudado de nome na versão 3.
Travar a versão é dizer: *funciona com esta aqui*.

### ⚠️ E quando a biblioteca não tem `__version__`?

Acontece bastante. Rode e veja o erro:

````python
%pip install qrcode --quiet
import qrcode
print(qrcode.__version__)
````

```
AttributeError: module 'qrcode' has no attribute '__version__'
```

Nem toda biblioteca expõe a versão assim. Mas o Python sabe, porque o pip anotou:

````python
from importlib.metadata import version
print(version("qrcode"))
````

**Guarde esse.** Funciona para qualquer pacote instalado, tendo `__version__` ou não.
