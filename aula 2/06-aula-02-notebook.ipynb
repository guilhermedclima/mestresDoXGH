# 1. O bug da tarefa de casa

Rode a célula abaixo e digite **2** e depois **2**.

Antes de rodar: escreva mentalmente o que você acha que vai aparecer.

````python
a = input("Primeiro número: ")
b = input("Segundo número: ")

print(a + b)
````

### Por que deu 22 e não 4?

Porque `input()` **sempre** devolve texto. Sempre — mesmo quando você digita um número.

E somar dois textos em Python não soma: **gruda**. Pelo mesmo motivo que `"casa" + "grande"` vira `"casagrande"`.

O computador não se confundiu. Ele fez exatamente o que foi pedido.

````python
print("casa" + "grande")
print("2" + "2")
````

### O conserto

`int()` pega um texto que parece número e transforma em número de verdade.

````python
a = int(input("Primeiro número: "))
b = int(input("Segundo número: "))

print(a + b)
````

# 2. Os quatro tipos básicos

O computador não guarda só o valor. Ele guarda o valor **e o que aquilo é**.

| Tipo | O que é | Exemplo |
|---|---|---|
| `str` | texto (sempre entre aspas) | `"Guilherme"` |
| `int` | número inteiro | `30` |
| `float` | número com casa decimal | `1.75` |
| `bool` | verdadeiro ou falso | `True` / `False` |

````python
nome = "Guilherme"
idade = 30
altura = 1.75
maior_de_idade = True

print(nome, idade, altura, maior_de_idade)
````

### `type()` — sua ferramenta de investigação

Toda vez que der um erro estranho, antes de sofrer: rode `type()` na variável.

Metade dos bugs de iniciante é o tipo não ser o que você achava que era.

````python
print(type(nome))
print(type(idade))
print(type(altura))
print(type(maior_de_idade))
````

### ⚠️ Três detalhes que confundem

1. `"30"` e `30` **não são a mesma coisa**. O primeiro é texto que parece número.
2. Em código o separador decimal é **ponto**, não vírgula. `1,75` não é um número.
3. `True` e `False` vão com maiúscula e **sem aspas**. Com aspas viram texto.

````python
print(type("30"))    # str — é texto!
print(type(30))      # int
print(type(True))    # bool
print(type("True"))  # str — com aspas virou texto
````

# 4. f-string: o jeito certo de montar texto

Primeiro, o jeito chato. Conte quantos `+` e quantos `str()` são necessários:

````python
nome = "Ana"
idade = 25

print("Oi, " + nome + "! Você tem " + str(idade) + " anos.")
````

Agora o jeito bom. O **`f`** antes das aspas libera as chaves `{}`:

````python
print(f"Oi, {nome}! Você tem {idade} anos.")
````

Tudo que estiver dentro de `{}` o Python calcula e coloca ali. Dá até para fazer conta:

````python
print(f"Ano que vem você faz {idade + 1}.")
print(f"A média entre 7 e 9 é {(7 + 9) / 2}")
````

### Controlando as casas decimais

`{valor:.2f}` mostra o número com **2 casas decimais**. Muito útil para dinheiro.

````python
preco = 19.9
print(f"Sem formatar: {preco}")
print(f"Formatado:   R$ {preco:.2f}")

print(f"Um terço com 3 casas: {1/3:.3f}")
````

> **A partir de hoje, f-string é o padrão da turma.** Some com metade dos erros de texto.

# 6. 💊 Pílula de IA: por que a máquina só entende número

Você passou a aula convertendo texto em número.

Agora pense: o ChatGPT recebe **texto**. Mas por dentro, computador só sabe fazer **conta**.

Como é que texto vira conta?

**Resposta: exatamente como hoje. Converte.**

### Passo 1 — toda letra já é um número por baixo

`ord()` mostra o número que representa cada caractere:

````python
print(ord("a"))
print(ord("b"))
print(ord("A"))
print(ord(" "))
````

Isso ainda **não é IA** — é só a tabela de caracteres, existe desde sempre.

Mas mostra que a ponte texto → número existe e é concreta.

### Passo 2 — quebrar a frase em pedaços e numerar

Não se preocupe em entender cada linha (parte disso é da aula 5).

Olhe só para o **resultado**: entrou frase, saiu lista de números.

````python
frase = "o gato subiu no telhado"

pedacos = frase.split()
print("Pedaços:", pedacos)

vocabulario = {"o": 1, "gato": 2, "subiu": 3, "no": 4, "telhado": 5}

numeros = [vocabulario[p] for p in pedacos]
print("Números:", numeros)
````

### As três frases que sustentam tudo que vem por aí

**1. Todo modelo de IA só come número.**
Texto, imagem e som viram número antes de entrar. Sem exceção.

**2. Esse processo tem nome: _tokenização_.**
Quebrar o texto em pedaços e trocar cada pedaço por um número. Você vai fazer isso de verdade na aula 5.

**3. Num modelo real, o número não é `1, 2, 3`.**
É uma lista de centenas de números que carrega *significado* — e que o modelo aprendeu sozinho lendo bilhões de frases.
Isso se chama **embedding**, e é o Módulo 4 do curso.
