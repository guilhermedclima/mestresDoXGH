# 2. Antes de decidir: a pergunta de sim ou não

O computador precisa de uma pergunta que tenha resposta **sim ou não**.

Em Python, essa resposta é um `bool` — o tipo que você viu na Aula 2.

````python
print(10 > 5)
print(10 < 5)
print(10 == 10)
print(10 != 10)
````

Repare: não imprimiu `10` nem `5`. Imprimiu **`True`** e **`False`**.

A comparação **vira um bool**. E é esse bool que o `if` vai olhar.

````python
print(type(10 > 5))
````

### Os seis comparadores

| Símbolo | Pergunta |
|---|---|
| `>` | é maior que? |
| `<` | é menor que? |
| `>=` | é maior ou igual? |
| `<=` | é menor ou igual? |
| `==` | é igual? |
| `!=` | é diferente? |

### ⚠️ O erro que pega todo mundo: `=` vs `==`

| | O que faz | Exemplo |
|---|---|---|
| `=` | **guarda** um valor | `idade = 18` |
| `==` | **pergunta** se são iguais | `idade == 18` |

> Um sinal de igual **manda**. Dois sinais de igual **perguntam**.

A célula abaixo está errada de propósito. Rode e leia o erro:

````python
idade = 20

if idade = 20:
    print("tem 20 anos")
````

```
SyntaxError: invalid syntax
```

Esse erro vai acontecer com você. Várias vezes.

**Regra:** viu `SyntaxError` numa linha de `if`? Procure o igual solitário antes de qualquer outra coisa.

Agora conserte a célula acima (coloque `==`) e rode de novo.

### Comparar texto também funciona

````python
print("banana" == "banana")
print("Banana" == "banana")   # maiúscula importa!
````

# 3. `if`, `else`, `elif`

## Só o `if`

````python
idade = 20

if idade >= 18:
    print("Pode dirigir")
````

Dois detalhes:

1. Os **dois pontos** `:` no fim da linha
2. O **espaço** (indentação) na linha de baixo — é ele que diz o que está *dentro* do `if`

### A indentação muda o programa

Rode as duas células abaixo. Repare que `idade = 15`.

````python
idade = 15

if idade >= 18:
    print("Pode dirigir")
    print("Boa viagem")     # DENTRO do if
````

````python
idade = 15

if idade >= 18:
    print("Pode dirigir")
print("Boa viagem")         # FORA do if
````

> O espaço não é enfeite. É a **única** coisa que separa esses dois programas.

## `else` — o caso contrário

````python
idade = 15

if idade >= 18:
    print("Pode dirigir")
else:
    print("Ainda não pode")
````

`else` não leva condição. Ele é **tudo que sobrou**.

## `elif` — mais de duas respostas

````python
idade = 15

if idade < 12:
    print("Criança")
elif idade < 18:
    print("Adolescente")
else:
    print("Adulto")
````

### 🔑 O detalhe que evita horas de bug

O Python lê de cima para baixo e **para no primeiro que der `True`**. Ele não testa o resto.

Por isso a segunda linha pode ser só `idade < 18`: se o programa chegou ali, já é garantido que não é menor que 12.

### 💥 A ordem importa — e o erro é silencioso

A célula abaixo tem uma criança de 5 anos. Rode e veja o que sai:

````python
idade = 5

if idade < 18:
    print("Adolescente")
elif idade < 12:
    print("Criança")      # nunca vai acontecer
````

Saiu **Adolescente** para uma criança de 5 anos.

E repare: **nenhum erro vermelho apareceu.**

> Esse é o bug mais perigoso que existe: o que não dá erro. O programa roda feliz e responde errado.

**Regra prática: do mais específico para o mais geral.**

Conserte a célula acima invertendo a ordem.

# 4. `and`, `or`, `not` — combinando condições

E se você precisar de **duas** coisas ao mesmo tempo?

````python
idade = 25
tem_habilitacao = True

if idade >= 18 and tem_habilitacao:
    print("Pode dirigir")
else:
    print("Não pode dirigir")
````

| Operador | É `True` quando | Em português |
|---|---|---|
| `and` | as **duas** condições são verdadeiras | "e" |
| `or` | **pelo menos uma** é verdadeira | "ou" |
| `not` | inverte o valor | "não" |

````python
print(True and True)
print(True and False)
print(True or False)
print(False or False)
print(not True)
````

### 💡 Dica que vale ouro

```python
# em vez de:
if tem_habilitacao == True:

# escreva:
if tem_habilitacao:
```

Se já é `bool`, não precisa comparar com `True`. Perguntar *"é verdade que é verdade?"* é redundante.

# 5. 🛠️ Mão na massa

## Parte A — calculadora de gastos com faixas

Agora a calculadora responde diferente dependendo da situação.

````python
salario = float(input("Salário: R$ "))
gastos = float(input("Gastos: R$ "))

sobra = salario - gastos
percentual = gastos / salario * 100

print()
if sobra < 0:
    print(f"⚠️  Você está R$ {abs(sobra):.2f} no vermelho.")
elif percentual > 80:
    print(f"Apertado: {percentual:.0f}% do salário vai embora.")
elif percentual > 50:
    print(f"Equilibrado: {percentual:.0f}% comprometido.")
else:
    print(f"Tranquilo: sobra R$ {sobra:.2f}, {100 - percentual:.0f}% do salário.")
````

`abs()` é novidade: tira o sinal de menos. `abs(-300)` dá `300`.

## Parte B — e se a pessoa digitar `abc`?

**A pergunta que ficou aberta na Aula 2.** A resposta chegou hoje:

# 6. 💊 Pílula de IA
## Um classificador é um `if` que ninguém escreveu

Hoje você aprendeu a **decidir**.

Um detector de spam decide: é spam ou não é.

Vamos fazer um.

### Etapa 1 — você já sabe fazer isso

Não se preocupe com o `def` e o `for` (aulas 6 e 4). Olhe o **`if` no meio** — ele é o classificador inteiro.

````python
def eh_spam(mensagem):
    suspeitas = ["promocao", "gratis", "clique", "premio"]
    for palavra in suspeitas:
        if palavra in mensagem:
            return True
    return False


print(eh_spam("promocao gratis clique aqui"))   # esperado: True
print(eh_spam("vamos almocar amanha"))          # esperado: False
````

### Etapa 2 — agora quebre o seu próprio classificador

````python
print(eh_spam("o curso e gratis para os alunos"))
````

Deu `True`. Uma mensagem honesta virou spam.

E agora? Você adiciona uma exceção. Aí quebra outra coisa. Adiciona outra.

> **A lista de regras nunca acaba** — e cada regra nova pode quebrar uma que já funcionava.

### Etapa 3 — e se o computador escrevesse as regras sozinho?

Abaixo: **16 mensagens de exemplo** com a resposta certa marcada.

Nenhuma regra. Nenhuma lista de palavras suspeitas. Só exemplos.

````python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

mensagens = [
    # --- spam ---
    "promocao imperdivel clique aqui agora",
    "voce ganhou um premio gratis resgate ja",
    "credito pre aprovado clique no link",
    "gratis so hoje aproveite a promocao",
    "clique aqui e ganhe dinheiro rapido",
    "ultima chance promocao relampago clique",
    "voce foi sorteado resgate seu premio",
    "emprestimo liberado clique no link agora",
    # --- normal ---
    "oi tudo bem vamos almocar amanha",
    "reuniao adiada para as 15h",
    "segue em anexo o relatorio do mes",
    "bom dia consegue revisar meu codigo hoje",
    "a aula de sabado comeca as 10h",
    "esqueci minha chave voce pode abrir",
    "parabens pelo trabalho ficou otimo",
    "vou chegar um pouco atrasado no almoco",
]

rotulos = ["spam"] * 8 + ["normal"] * 8

# as 3 linhas que treinam o modelo
vetor = CountVectorizer()
X = vetor.fit_transform(mensagens)
modelo = MultinomialNB().fit(X, rotulos)

print("Modelo treinado com", len(mensagens), "exemplos.")
````

Agora testando com frases que ele **nunca viu**:

````python
def classificar(frase):
    x = vetor.transform([frase])
    resposta = modelo.predict(x)[0]
    confianca = modelo.predict_proba(x).max()
    print(f"{resposta:7} ({confianca:.0%})  <- {frase}")


classificar("promocao gratis clique aqui")
classificar("vamos almocar amanha")
classificar("clique no link e ganhe dinheiro")
classificar("segue o relatorio em anexo")
classificar("ultima chance para resgatar seu premio gratis")
classificar("bom dia tudo bem")
````

Repare no que **não** existe aqui:

- Nenhuma regra escrita
- Nenhuma lista de palavras suspeitas

Só 16 exemplos com a resposta certa. Ele descobriu o resto sozinho.

### Etapa 4 — o modelo também erra (e isso é importante)

````python
classificar("o curso e gratis para os alunos")
classificar("voce pode me mandar o link da reuniao")
````

Errou igual à regra manual.

**Mas olhe a confiança: 61% e 55%.**

Ele está em cima do muro — e *sabe* disso. A regra manual dizia `True` com certeza absoluta.

> Um modelo que sabe que está inseguro é muito mais útil que uma regra convicta e errada.

Em sistema sério, caso com confiança baixa vai para um humano olhar.

### Etapa 5 — o desfecho

Vamos consertar. Mas repare **como**: sem escrever nenhuma regra nova.

Só adicionando 8 exemplos de mensagens normais que usam "gratis", "clique" e "link".

````python
novos_exemplos = [
    "o curso e gratis para os alunos",
    "a entrada do evento e gratis hoje",
    "clique no botao verde do sistema",
    "voce pode me mandar o link da reuniao",
    "o estacionamento e gratis no sabado",
    "clique em salvar antes de fechar",
    "me manda o link do documento por favor",
    "a palestra e gratis para estudantes",
]

mensagens2 = mensagens + novos_exemplos
rotulos2 = rotulos + ["normal"] * 8

vetor = CountVectorizer()
X2 = vetor.fit_transform(mensagens2)
modelo = MultinomialNB().fit(X2, rotulos2)

print("Agora com", len(mensagens2), "exemplos.\n")

classificar("o curso e gratis para os alunos")
classificar("voce pode me mandar o link da reuniao")
classificar("a entrada do evento e gratis")
classificar("promocao gratis clique aqui")     # continua acertando o spam
classificar("vamos almocar amanha")
````

## As três frases para guardar

**1. O `if` continua lá dentro.**
O modelo decide comparando números com limiares. A diferença é que **você não escreveu os limiares** — ele calculou olhando exemplos.

**2. Por isso dado vale mais que código em IA.**
Você corrigiu o modelo sem escrever uma regra. Deu mais exemplos. Essa é a diferença entre programação normal e machine learning.

**3. O modelo diz o quanto está inseguro.**
Aquele 61% é informação valiosa, não defeito.

---

> Um classificador de IA é um `if` gigante cujos limiares ninguém escreveu à mão.
