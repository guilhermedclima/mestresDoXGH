# 2. O `for`: para cada coisa, faça isso

Leia em português, da esquerda para a direita:

```
for  aluno  in  alunos:
para cada aluno dentro de alunos, faça:
```

````python
alunos = ["Ana", "Bruno", "Carla"]

for aluno in alunos:
    print(f"Olá, {aluno}!")
````

Aquele `alunos` entre colchetes é uma **lista** — por enquanto pense nela como um saco de coisas.
A gente destrincha lista na Aula 5. Hoje ela é só o que o `for` percorre.

**Vocabulário de hoje** (anota, porque é isso que faz você conseguir pesquisar sozinho):

| Palavra | O que é |
|---|---|
| **laço** (ou *loop*) | o bloco que repete |
| **iteração** | uma volta do laço |
| **variável do laço** | o nome que muda a cada volta (`aluno`) |
| **corpo do laço** | as linhas indentadas, que rodam a cada volta |

### O nome da variável do laço é seu

`aluno` não é palavra mágica. Podia ser qualquer coisa — e o programa faz exatamente o mesmo:

````python
for x in alunos:
    print(f"Olá, {x}!")
````

### A indentação de novo (a mesma da Aula 3)

O que está indentado é o **corpo do laço** — roda a cada volta.
O que não está, roda **uma vez só**, depois que o laço acabar.

Rode as duas células abaixo e compare a saída.

````python
for aluno in alunos:
    print(aluno)
    print("--- fim ---")
````

````python
for aluno in alunos:
    print(aluno)
print("--- fim ---")
````

# 3. `range`: quando você quer números, não coisas

Nem sempre você tem uma lista. Às vezes você só quer *repetir N vezes*.

````python
for i in range(5):
    print(i)
````

Repare em **duas coisas** que pegam todo mundo:

1. Começou no **0**, não no 1.
2. Foram **5 números** — e o último é o **4**.

O número que você escreve dentro do `range` é a **quantidade de voltas**, não o último valor.

````python
print(list(range(5)))        # 5 números, começando do zero
print(list(range(1, 6)))     # do 1 até o 6 — SEM incluir o 6
print(list(range(0, 10, 2))) # de 2 em 2
````

### 💥 Quebrando de propósito nº 1

Você quer contar de 1 a 5. Escreve o óbvio:

````python
for i in range(1, 5):
    print(i)
````

Saiu `1 2 3 4`. **Cadê o 5?**

`range(1, 5)` para **antes** do 5. Para chegar no 5, você pede `range(1, 6)`.

> Isso vai te morder pelo menos umas dez vezes. Quando o seu laço rodar uma volta a menos
> do que devia, o `range` é o primeiro lugar para olhar.

E mais uma pegadinha: `range` sozinho não mostra os números.

````python
print(range(5))
````

Ele é uma *receita* para gerar números, não os números prontos.

# 4. O acumulador — a ideia mais útil da aula

Repetir só vale a pena se alguma coisa **sobreviver** de uma volta para a outra.
Essa coisa é o acumulador.

````python
gastos = [300, 150, 80, 420]

total = 0
for gasto in gastos:
    total = total + gasto

print(f"Total gasto: R$ {total}")
````

Olhe a linha do meio com carinho:

```python
total = total + gasto
```

Em matemática isso seria absurdo (`x = x + 5`?). Em Python não é uma equação —
lembra da Aula 3? **Um sinal de igual manda.** Ele está mandando:

> *pegue o `total` de agora, some o `gasto` desta volta, e guarde de volta no `total`.*

Como isso é comum demais, existe um atalho — e é o que você vai ver em código de verdade:

````python
total = 0
for gasto in gastos:
    total += gasto      # exatamente a mesma coisa

print(total)
````

### Contador: acumulador que só conta

E aqui as duas ferramentas se encontram — o `if` de sábado passado **dentro** do `for` de hoje:

````python
notas = [4, 8, 6, 9, 3, 7, 10, 5]

aprovados = 0
for nota in notas:
    if nota >= 7:
        aprovados += 1

print(f"{aprovados} aprovados de {len(notas)}")
````

`len()` é novidade: devolve **quantos itens** tem na lista.

Repare na indentação em **dois níveis**: o `if` está dentro do `for`, e o `aprovados += 1`
está dentro do `if`. Cada nível é mais quatro espaços.

### 💥 Quebrando de propósito nº 2

Um erro de **um caractere de indentação** — e nenhum erro vermelho aparece:

````python
total = 0
for gasto in gastos:
    total = 0        # ← dentro do laço! zera a cada volta
    total += gasto

print(total)
````

Deu **420**. O certo era **950**.

420 é o último gasto da lista — porque o `total` foi zerado logo antes.

> É o mesmo tipo de bug da ordem dos `elif` da Aula 3: **o programa roda feliz e responde errado.**
> O acumulador nasce **antes** do laço. Sempre.

# 5. `while`: quando você não sabe quantas voltas

O `for` é para quando você **sabe** o número de repetições: cada aluno da lista, 10 da tabuada.

O `while` é para quando você **não sabe**: "continue até dar certo".

Leia em português: *enquanto a condição for verdadeira, faça.*

````python
n = 5
while n > 0:
    print(n)
    n = n - 1

print("Fogo!")
````

Todo `while` tem **três partes**. Se faltar uma, quebra:

1. **Antes:** alguma coisa que a condição consiga olhar (`n = 5`)
2. **A condição:** a pergunta de sim ou não (`n > 0`)
3. **Dentro:** alguma coisa que **muda** a condição (`n = n - 1`)

A parte 3 é a que todo mundo esquece. Quando esquece, acontece o seguinte.

### 💥 Quebrando de propósito nº 3 — o loop infinito

⚠️ **A célula abaixo NÃO PARA sozinha.** Rode ela de propósito e depois clique no
botão **⏹ (parar)** à esquerda da célula, no Colab.

Saber parar um programa travado é habilidade, não acidente. Todo mundo vai fazer isso.

````python
# ⚠️ RODE E DEPOIS APERTE O BOTÃO DE PARAR (⏹)
n = 1
while n < 5:
    print(n)
    # faltou o n = n + 1 — a condição nunca deixa de ser verdadeira
````

Nenhum erro vermelho. O programa está fazendo **exatamente** o que você mandou:
repetir enquanto `n < 5`. E o `n` nunca muda.

### `break`: a saída de emergência

`break` interrompe o laço na hora, esteja onde estiver.

````python
total = 0
numero = 0

while True:              # "para sempre"
    numero += 1
    total += numero
    if total > 100:
        break            # ← a única saída

print(f"Passamos de 100 no {numero} (total {total})")
````

### `for` ou `while`?

| | Use quando |
|---|---|
| **`for`** | você sabe quantas voltas: uma por item da lista, 10 da tabuada, `range(200)` |
| **`while`** | você não sabe: até o usuário acertar, até o erro ficar pequeno, até acabar o arquivo |

Na dúvida, comece pelo `for` — ele não trava.

# 6. Mão na massa

## Parte A — a calculadora, agora para a equipe inteira

A calculadora da Aula 3 atendia **uma** pessoa. Com um `for`, ela atende quantas você quiser.

````python
salarios = [2800, 4500, 1900, 7200, 3100, 5600]

soma = 0
acima = 0

for salario in salarios:
    soma += salario
    if salario > 3000:
        acima += 1

print(f"Folha total:    R$ {soma}")
print(f"Média:          R$ {soma / len(salarios):.2f}")
print(f"Acima de 3000:  {acima} de {len(salarios)}")
````

## Parte B — os 200 `elif`, a resposta completa

A pergunta tem **duas** respostas, e a certa depende das faixas.

### Resposta 1 — se as faixas seguem uma regra, você **calcula**

Faixas de mil em mil não precisam de `if` nenhum. Precisam de uma divisão.
`//` divide e joga fora o resto.

````python
for valor in [850, 2400, 4200, 7300, 25000]:
    faixa = valor // 1000
    print(f"R$ {valor} está na faixa {faixa}")
````

### Resposta 2 — se as faixas são arbitrárias, você **guarda a tabela e percorre**

````python
limites = [1000, 3000, 5000, 10000]

for valor in [850, 2400, 4200, 7300, 25000]:
    faixa = "até 1000"
    for limite in limites:
        if valor >= limite:
            faixa = f"acima de {limite}"
    print(f"R$ {valor} → {faixa}")
````

Agora o ponto, e ele é a aula toda em uma frase:

> **Para ter 200 faixas, a lista `limites` cresce. O código não muda uma linha.**

Isso é a diferença entre um programa que você escreve e um programa que você **alimenta**.

# 💊 Pílula de IA — treinar é repetir

## Etapa 1 — o problema

Você anotou quantas horas cada aluno estudou e que nota tirou:

| Horas | Nota |
|---|---|
| 1 | 2 |
| 2 | 4 |
| 3 | 5 |
| 4 | 7 |
| 5 | 8 |
| 6 | 10 |

Você quer prever a nota de quem estudou 3 horas e meia. O palpite mais simples possível:

```
nota = peso × horas
```

Só falta **um número**: o `peso`. Vamos chutar.

````python
horas = [1, 2, 3, 4, 5, 6]
notas = [2, 4, 5, 7, 8, 10]

peso = 0.5   # chute

for i in range(len(horas)):
    print(f"{horas[i]}h → previsto {peso * horas[i]:.1f} | real {notas[i]}")
````

## Etapa 2 — o erro é só uma conta

Se o erro é um **número**, dá para comparar dois chutes e saber qual é melhor.

Para cada aluno: pegue a diferença entre o previsto e o real, eleve ao quadrado
(para o sinal não importar) e some tudo. Um `for` com acumulador — o que você aprendeu hoje.

````python
def erro_total(peso):
    erro = 0
    for i in range(len(horas)):
        previsto = peso * horas[i]
        erro += (previsto - notas[i]) ** 2
    return erro


for chute in [0.0, 0.5, 1.0, 2.0, 3.0]:
    print(f"peso {chute:.1f} → erro {erro_total(chute):7.2f}")
````

Repare no formato do erro: ele **cai, cai... e em algum ponto volta a subir**. Existe um peso
melhor que todos os outros, em algum lugar entre 1.0 e 3.0. Mas **onde**, exatamente?

Achar esse número na mão, chutando, até dá — com **um** número. Um modelo de verdade tem
milhões deles. Ninguém chuta milhões de números.

*(Duas coisas nessas células são de aulas futuras: `def` é da Aula 6 e `horas[i]` — pegar o
item da posição `i` — é da Aula 5. Hoje, olhe só para os laços.)*

## Etapa 3 — o laço que aprende sozinho

Em vez de chutar, o computador faz assim: começa em qualquer lugar, olha para que lado
o erro diminui, **anda um passinho para esse lado** — e repete.

````python
peso = 0.0
taxa = 0.001      # o tamanho do passinho
historico = []

for passo in range(50):
    ajuste = 0
    for i in range(len(horas)):
        previsto = peso * horas[i]
        ajuste += 2 * (previsto - notas[i]) * horas[i]

    peso = peso - taxa * ajuste          # ← o acumulador de hoje
    historico.append(erro_total(peso))

    if passo % 10 == 0 or passo == 49:
        print(f"passo {passo:2d} | peso {peso:.3f} | erro {erro_total(peso):7.2f}")
````

**Erro 258 no começo. Erro 0.76 no fim.** Ninguém escreveu o `1.681`.

E agora olhe para o que esse código **é**:

- um `for` com `range`
- um acumulador (`ajuste += ...`)
- uma linha do tipo `peso = peso - alguma coisa`

**Você aprendeu as três coisas hoje de manhã.** Isso é treinar um modelo.

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 3.5))
plt.plot(historico, linewidth=2)
plt.title("O erro caindo a cada volta do laço")
plt.xlabel("passo (uma volta do for)")
plt.ylabel("erro")
plt.grid(alpha=.3)
plt.show()
````

## Etapa 4 — 💥 e se o passo for grande demais?

O `taxa = 0.001` é o tamanho do passinho. Parece um detalhe. Não é.

````python
peso = 0.0
taxa = 0.02       # vinte vezes maior

for passo in range(8):
    ajuste = 0
    for i in range(len(horas)):
        ajuste += 2 * (peso * horas[i] - notas[i]) * horas[i]
    peso = peso - taxa * ajuste
    print(f"passo {passo} | peso {peso:12.2f} | erro {erro_total(peso):18.2f}")
````

O erro **explodiu**: de 258 para mais de um bilhão em oito voltas.

O passo foi tão grande que ele passou do ponto certo, foi parar do outro lado — mais longe
ainda — e da próxima vez passou mais ainda.

Isso tem nome e é o botão mais importante de quem treina modelo: **taxa de aprendizado**
(*learning rate*). Passo pequeno demais: demora uma eternidade. Grande demais: nunca chega.

Quando você ler que "o treino divergiu", é exatamente esta célula.

## Etapa 5 — a escala

Esse laço rodou **50 vezes** ajustando **1 número**.

Treinar um modelo de linguagem é **o mesmo laço**, rodando bilhões de vezes, ajustando
bilhões de números. A conta de "para que lado empurrar" fica mais complicada — na faculdade
isso se chama **gradiente descendente** — mas a estrutura é essa, e você acabou de escrevê-la.

---

### O que fica da pílula de hoje

1. **Treinar não é mágica.** É um `for` que chuta, mede o erro e corrige. Muitas vezes.
2. **O modelo não sabe a resposta** — ele descobre errando de propósito e ajustando.
3. **Ninguém escreveu o `1.681`.** Igual à semana passada: os dados escreveram.
