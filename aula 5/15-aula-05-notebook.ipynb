# 1. A pergunta que ficou de sábado passado

Você usou lista a aula inteira e eu nunca expliquei o que ela é. E o `for` só sabe uma coisa:
**andar do começo ao fim, um por um.**

Aí eu perguntei: e se eu quiser o **maior** salário da lista?

Com o que você já sabe, dá — e a resposta é bonita:

````python
salarios = [2800, 4500, 1900, 7200, 3100, 5600]

maior = salarios[0]
for salario in salarios:
    if salario > maior:
        maior = salario

print(maior)
````

Isso é o acumulador da Aula 4 de novo — só que em vez de somar, ele guarda **o maior até agora**.

Funciona. Mas hoje as coisas ficam assim:

````python
print(max(salarios))                    # o maior
print(min(salarios))                    # o menor
print(sum(salarios))                    # a soma
print(salarios[2])                      # o terceiro
print(sorted(salarios, reverse=True))   # ordenados do maior para o menor
````

Cinco perguntas, cinco linhas.

> **Não jogue fora o laço de cima.** Saber fazer na mão é o que te deixa entender o que o `max()`
> faz por dentro — e é o que te salva no dia em que a pergunta for esquisita demais para ter
> uma função pronta.

# 2. Índice: pedir um item pela posição

````python
alunos = ["Ana", "Bruno", "Carla", "Diego", "Elis"]

print(alunos[0])    # o primeiro
print(alunos[2])    # o terceiro
print(alunos[-1])   # o último
````

Duas coisas para gravar:

1. **A contagem começa no zero.** O primeiro é `[0]`, o terceiro é `[2]`.
2. **Índice negativo conta de trás para frente.** `[-1]` é o último, `[-2]` o penúltimo.

```
   Ana   Bruno   Carla   Diego   Elis
    0      1       2       3       4
   -5     -4      -3      -2      -1
```

### 💥 Quebrando de propósito nº 1

A lista tem 5 nomes. Peça o quinto — do jeito que parece certo:

````python
print(alunos[5])
````

```
IndexError: list index out of range
```

São 5 itens, mas o último índice é **4**. Igual ao `range` da semana passada: a contagem
começa no zero, então o último é sempre `len(lista) - 1`.

> `IndexError` quer dizer sempre a mesma coisa: **você pediu uma posição que não existe.**
> Na dúvida, imprima `len(lista)` antes.

````python
print(len(alunos))        # quantos itens
print(alunos[len(alunos) - 1])   # o último, na marra
print(alunos[-1])                # o último, do jeito civilizado
````

# 3. Fatiamento: pedir um pedaço

Dois números separados por `:` — começo e fim.

````python
print(alunos[1:3])    # do 1 até o 3, sem incluir o 3
print(alunos[:2])     # do começo até o 2
print(alunos[2:])     # do 2 até o fim
print(alunos[-2:])    # os dois últimos
````

**O fim não entra.** É a mesma regra do `range(1, 5)` que te mordeu semana passada — o Python
é teimosamente consistente nisso, e isso é bom: uma regra só para lembrar.

Um jeito de nunca mais errar: `[1:3]` significa *"comece no 1 e pegue até chegar no 3"*.
Quando chega no 3, para. Não pega o 3.

# 4. Mexer na lista

Lista não é só para ler. Ela muda.

````python
turma = ["Ana", "Bruno"]

turma.append("Carla")      # põe no fim
print(turma)

turma.insert(1, "Zeca")    # põe na posição 1, empurrando o resto
print(turma)

turma.remove("Bruno")      # tira pelo VALOR, não pela posição
print(turma)
````

Note o `.` antes do nome. `turma.append(...)` se lê *"turma, acrescente"*. São coisas que a
**lista sabe fazer com ela mesma**. Isso tem nome: **método**. Vocabulário do dia.

E duas perguntas que já apareceram antes:

````python
print(len(turma))        # quantos
print("Ana" in turma)    # está aí dentro?  ← o mesmo `in` da Aula 3
````

# 5. Ordenar — e a armadilha

`sorted()` devolve uma lista nova, ordenada. A original fica intacta.

````python
notas = [7, 4, 9, 6, 10]

print(sorted(notas))                 # crescente
print(sorted(notas, reverse=True))   # decrescente
print(notas)                         # a original NÃO mudou
````

Existe também o método `.sort()`, que ordena **a própria lista**, no lugar:

````python
notas.sort()
print(notas)     # agora a original mudou
````

### 💥 Quebrando de propósito nº 2

Parece a coisa mais natural do mundo escrever isto:

````python
notas = [7, 4, 9]
notas = notas.sort()

print(notas)
````

```
None
```

**A lista sumiu.** E se você tentar usar:

```python
len(notas)
→ TypeError: object of type 'NoneType' has no len()
```

Por quê: `.sort()` ordena a lista e **não devolve nada**. Ao escrever `notas = notas.sort()`,
você jogou esse nada por cima da sua lista.

> A regra que resolve isso de vez:
> - `sorted(lista)` → **devolve** uma lista nova. Você precisa guardar o resultado.
> - `lista.sort()` → **muda** a lista. Você NÃO guarda nada.
>
> Nunca as duas coisas ao mesmo tempo.

# 6. A virada da aula: string também é uma sequência

Tudo que você fez com lista funciona com texto. **Uma string é uma sequência de caracteres.**

````python
nome = "Guilherme"

print(nome[0])      # G
print(nome[-1])     # e
print(nome[0:4])    # Guil
print(len(nome))    # 9
````

Exatamente os mesmos colchetes, a mesma contagem do zero, o mesmo fim que não entra.

E strings têm os métodos delas:

````python
print("guilherme".upper())
print("GUILHERME".lower())
print("   com espaço   ".strip() + "|")   # tira espaço das pontas
print("XGH".replace("H", "!"))
````

### 💥 Quebrando de propósito nº 3

Se string é uma sequência igual à lista, então dá para trocar uma letra, certo?

````python
nome = "guilherme"
nome[0] = "G"
````

```
TypeError: 'str' object does not support item assignment
```

**String é imutável.** Ela não muda — nunca. Você só cria uma nova.

E é por isso que isto não faz o que parece:

````python
nome = "guilherme"
nome.upper()
print(nome)
````

Saiu `guilherme`, minúsculo. O `.upper()` **devolveu** a versão maiúscula e você jogou fora.
Para ficar, tem que guardar:

````python
nome = nome.upper()
print(nome)
````

### A regra que junta os dois erros de hoje

| | O que faz | Você guarda? |
|---|---|---|
| `lista.append(x)`, `lista.sort()` | **muda** a lista | não |
| `sorted(lista)`, `texto.upper()`, `texto.strip()` | **devolve** algo novo | **sim** |

Lista muda. String nunca muda.

Quando um método "não funcionou", pergunte primeiro: *ele muda ou devolve?*

# 8. Mão na massa

## Parte A — o boletim da turma

````python
nomes = ["Ana", "Bruno", "Carla", "Diego", "Elis"]
notas = [8.5, 4.0, 9.5, 6.0, 7.5]

print(f"Turma: {len(nomes)} alunos")
print(f"Média da turma: {sum(notas) / len(notas):.2f}")
print(f"Maior nota: {max(notas)}  |  Menor: {min(notas)}")
print()

aprovados = 0
for i in range(len(nomes)):
    if notas[i] >= 7:
        aprovados += 1
        print(f"{nomes[i]:8s} {notas[i]:5.1f}  aprovado")
    else:
        print(f"{nomes[i]:8s} {notas[i]:5.1f}  reprovado")

print()
print(f"Aprovados: {aprovados} de {len(nomes)}")
print("Notas ordenadas:", sorted(notas, reverse=True))
print("Top 3:", sorted(notas, reverse=True)[:3])
````

O `range(len(nomes))` é o truque para andar em **duas listas ao mesmo tempo**: o `i` serve de
posição nas duas. Guarde — ele aparece muito.

## Parte B — brincando com a frase

````python
frase = "programar e a arte de resolver problemas com codigo"
palavras = frase.split()

# a maior palavra — o mesmo padrão do "maior salário" lá do começo
maior = palavras[0]
for p in palavras:
    if len(p) > len(maior):
        maior = p
print("Maior palavra:", maior, f"({len(maior)} letras)")

# as iniciais — acumulador da Aula 4, mas com texto
iniciais = ""
for p in palavras:
    iniciais += p[0].upper()
print("Iniciais:", iniciais)

# a frase de trás para frente
print("Invertida:", " ".join(palavras[::-1]))
````

Repare no `iniciais += p[0].upper()`: é o acumulador de sábado passado, **só que somando texto
em vez de número**. A ideia é a mesma; mudou o tipo.

E o `[::-1]` é um fatiamento com passo `-1` — anda de trás para frente. Vale guardar como truque.

# 💊 Pílula de IA — você acabou de escrever um tokenizador

## Etapa 1 — o `.split()` é um tokenizador

Um modelo de linguagem não lê frase. Ele lê **pedaços**. Partir texto em pedaços tem nome:
**tokenização**. E você fez isso hoje:

````python
frase = "o curso e gratis para os alunos"
print(frase.split())
````

Pronto. Tokenizador. É esse o primeiro passo de qualquer modelo que lida com texto.

## Etapa 2 — agora quebre o seu tokenizador

Duas frases que deviam ser quase iguais:

````python
print("Gato!".split())
print("gato".split())
print("gato,".split())
````

Para o computador, `'Gato!'`, `'gato'` e `'gato,'` são **três coisas completamente diferentes**.

E tem pior. Se cada palavra vira um item do vocabulário, quantos itens você precisa?

- português tem mais de 400 mil palavras
- mais os plurais, as conjugações, os nomes próprios
- mais `"aaaaah"`, `"kkkkk"`, `"neymarzete"`, e tudo que a internet inventar amanhã

**O vocabulário nunca fecha.** E uma palavra que o modelo nunca viu vira um buraco: ele não
tem nada para colocar ali.

## Etapa 3 — a saída: não parta em palavras, parta em **pedaços**

A ideia é essa: comece com as letras soltas e vá **juntando os pares que mais aparecem**.
Um `for`, um contador, um `max` — tudo de hoje e de sábado passado.

````python
corpus = """
o gato dorme o gato come o gato corre
o rato dorme o rato come o rato corre
a gata dorme a gata come a gata corre
""".split()

palavras = []
for p in corpus:
    palavras.append(list(p))     # cada palavra vira lista de letras

print("Vocabulário inicial:", sorted(set("".join(corpus))))
print()

for passo in range(6):
    # conta cada par de vizinhos
    contagem = {}
    for p in palavras:
        for i in range(len(p) - 1):
            par = (p[i], p[i + 1])
            contagem[par] = contagem.get(par, 0) + 1

    melhor = max(contagem, key=contagem.get)
    novo = melhor[0] + melhor[1]
    print(f"passo {passo + 1}: junta {melhor[0]!r} + {melhor[1]!r} -> {novo!r}   (aparecia {contagem[melhor]}x)")

    # aplica a junção em todas as palavras
    novas = []
    for p in palavras:
        q, i = [], 0
        while i < len(p):
            if i < len(p) - 1 and (p[i], p[i + 1]) == melhor:
                q.append(novo)
                i += 2
            else:
                q.append(p[i])
                i += 1
        novas.append(q)
    palavras = novas
````

````python
vistas = []
for original, pedacos in zip(corpus, palavras):
    if original not in vistas:
        vistas.append(original)
        print(f"{original:8s} -> {pedacos}")
````

Olhe o que ele descobriu **sozinho**, sem ninguém ensinar português para ele:

- juntou `a`+`t` → `at`, depois `g`+`at` → `gat`, depois `gat`+`o` → `gato`
- e `gata` ficou `['gat', 'a']` — **compartilhando o `gat` com `gato`**

Ninguém disse que `gato` e `gata` têm a mesma raiz. Ele viu que aqueles caracteres andavam
juntos e economizou. Isso se chama **BPE** (*byte pair encoding*), e é exatamente o algoritmo
que o GPT usa. Você acabou de rodar a versão de brinquedo dele.

## Etapa 4 — o tokenizador de verdade

Este é o mesmo que roda por trás do ChatGPT:

````python
%pip install -q tiktoken
````

````python
import tiktoken
enc = tiktoken.get_encoding("cl100k_base")

for palavra in ["gato", "cachorro", "sábado", "programação"]:
    ids = enc.encode(palavra)
    pedacos = []
    for i in ids:
        pedacos.append(enc.decode([i]))
    print(f"{palavra:14s} -> {len(ids)} tokens: {pedacos}")
````

`gato` — uma palavra curtíssima em português — vira **dois** tokens: `'g'` e `'ato'`.

Por quê? Porque esse tokenizador foi treinado quase todo em inglês. Ele nunca viu português
suficiente para achar que valia a pena guardar `gato` inteiro.

E isso tem preço. Literalmente:

````python
pt = "O curso de programação começa às dez horas de sábado"
en = "The programming course starts at ten o'clock on Saturday"

for lingua, texto in [("PT", pt), ("EN", en)]:
    ids = enc.encode(texto)
    print(f"{lingua}: {len(texto.split()):2d} palavras -> {len(ids):2d} tokens")
````

A **mesma frase**: 12 tokens em português, 10 em inglês.

## Etapa 5 — por que isso importa para você

1. **A API cobra por token, não por palavra.** O mesmo texto em português custa mais caro.
2. **O limite de contexto é em token.** Quando você lê "128 mil tokens", isso é bem menos texto
   em português do que em inglês.
3. **O modelo nunca vê a sua letra.** Ele vê esses pedaços virarem números — e aí vale a Pílula
   da Aula 2: *todo modelo só entende número*.
