# 1. A pergunta que ficou de sábado passado

Olhe o que você escreveu três vezes na semana passada:

````python
notas = [8.5, 4.0, 9.5, 6.0, 7.5]
salarios = [2800, 4500, 1900, 7200, 3100, 5600]
palavras = "programar e a arte de resolver problemas".split()

# no boletim
maior = notas[0]
for item in notas:
    if item > maior:
        maior = item
print(maior)

# no relatório de salários
maior = salarios[0]
for item in salarios:
    if item > maior:
        maior = item
print(maior)

# no contador de palavras
maior = palavras[0]
for item in palavras:
    if len(item) > len(maior):
        maior = item
print(maior)
````

**O que muda de um bloco para o outro?** Praticamente nada — só a lista.

Quando só um pedaço muda, você já sabe o que fazer: foi exatamente isso que o `for` resolveu na
Aula 4. Só que agora o que se repete não é uma linha, é um **bloco inteiro**.

E a ferramenta para isso é esta:

````python
def maior_de(lista):
    maior = lista[0]
    for item in lista:
        if item > maior:
            maior = item
    return maior


print(maior_de(notas))
print(maior_de(salarios))
````

Três blocos viraram um. E agora vem a parte que interessa de verdade.

Eu perguntei: **e no dia em que a regra mudar?** Aprovado passa a ser 6, e não 7 —
você vai lembrar dos três lugares?

````python
def situacao(nota, corte=7):
    if nota >= corte:
        return "aprovado"
    else:
        return "reprovado"


print(situacao(8.5))
print(situacao(6.0))
print(situacao(6.0, corte=6))    # a regra nova, sem mexer na função
````

A regra mora **num lugar só**. Se mudar, muda ali. Não tem terceiro lugar esquecido para te
morder daqui a três meses.

> Esse é o ponto da aula inteira, e vale muito mais que a sintaxe:
> **função não é para escrever menos. É para ter um lugar só onde mexer.**

# 2. Anatomia de uma função

````python
def saudacao(nome):
    return f"Olá, {nome}! Bem-vindo ao XGH."


print(saudacao("Ana"))
print(saudacao("Bruno"))
````

```
def  saudacao (nome) :
 │       │       │   └── dois pontos, igual ao if e ao for
 │       │       └────── PARÂMETRO: o que entra
 │       └────────────── NOME: como você vai chamar
 └────────────────────── "define"

    return f"Olá, {nome}!"
      └── o que SAI
```

Duas coisas que você já conhece: os **dois pontos** no fim e a **indentação** marcando o corpo.
São as mesmas do `if` e do `for`. O Python é consistente — de novo.

**Definir não executa.** A célula acima só rodou porque tem o `print(saudacao("Ana"))` no fim.
Sem a chamada, o `def` sozinho não faz nada — é uma receita guardada.

# 3. `return` não é `print` — e essa confusão custa caro

### 💥 Quebrando de propósito nº 1

````python
def dobro_errado(x):
    print(x * 2)

def dobro_certo(x):
    return x * 2


a = dobro_errado(5)
b = dobro_certo(5)

print("com print :", a)
print("com return:", b)
````

As duas **mostraram** 10 na tela. Mas quando você guarda o resultado:

- `dobro_certo(5)` te **entregou** o 10 — dá para guardar, somar, usar
- `dobro_errado(5)` só **mostrou** e devolveu `None`

E aí:

````python
print(dobro_errado(5) + 1)
````

```
TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'
```

**Aquele `None` de novo.** É o mesmo da Aula 5, quando `notas = notas.sort()` apagou a lista.

> **Função sem `return` devolve `None`.** Sempre.
>
> `print` é para o **humano** ver. `return` é para o **programa** usar.
> Quase toda função que você escrever deve ter `return`, não `print`.

E `return` também **encerra** a função na hora — o que vem depois não roda:

````python
def teste(n):
    if n > 10:
        return "grande"
    return "pequeno"


print(teste(50))
print(teste(3))
````

# 4. Escopo: o que nasce dentro, morre dentro

### 💥 Quebrando de propósito nº 2

````python
def calcular():
    resultado = 42
    return resultado


calcular()
print(resultado)
````

```
NameError: name 'resultado' is not defined
```

A função rodou. O `resultado` existiu. E sumiu quando ela acabou.

> Variável criada dentro de uma função **só existe lá dentro**. É por isso que o `return` importa
> tanto: ele é a única porta de saída.

Para usar o valor, guarde o que a função devolveu:

````python
resultado = calcular()
print(resultado)
````

Isso não é limitação — **é o que faz função ser útil.** Você pode escrever `maior_de()` sem se
preocupar se já existe um `maior` em algum outro lugar do programa. Cada função é uma sala com a
porta fechada.

# 🎮 Projeto 1 — Jogo de adivinhação

O primeiro programa de verdade de vocês. E ele fica bom justamente porque é feito de
**funções pequenas, cada uma com um trabalho só**.

````python
import random


def sortear():
    return random.randint(1, 100)


def comparar(chute, segredo):
    if chute < segredo:
        return "maior"
    elif chute > segredo:
        return "menor"
    else:
        return "acertou"


def avaliar(tentativas):
    if tentativas <= 5:
        return "Excelente!"
    elif tentativas <= 8:
        return "Muito bom."
    else:
        return "Conseguiu!"


print(comparar(30, 50))
print(comparar(70, 50))
print(comparar(50, 50))
print(avaliar(4), avaliar(7), avaliar(12))
````

Agora o jogo em si. Repare que ele **se lê como a explicação do jogo** — porque cada linha chama
uma função com nome de gente:

````python
def jogar():
    segredo = sortear()
    tentativas = 0

    print("Pensei num número de 1 a 100.")
    while True:
        chute = int(input("Seu chute: "))
        tentativas += 1
        resposta = comparar(chute, segredo)

        if resposta == "acertou":
            print(f"Acertou em {tentativas} tentativas! {avaliar(tentativas)}")
            break
        else:
            print(f"O número é {resposta}.")


jogar()
````

## E agora a parte divertida: o computador jogando contra si mesmo

Como o jogo é feito de funções, dá para **trocar o jogador** sem mexer em nada do resto.
Este aqui sempre chuta o meio do intervalo que sobrou:

````python
def jogar_robo(mostrar=True):
    segredo = sortear()
    baixo, alto = 1, 100
    tentativas = 0

    while True:
        chute = (baixo + alto) // 2
        tentativas += 1
        resposta = comparar(chute, segredo)

        if mostrar:
            print(f"  tentativa {tentativas}: chutei {chute} -> {resposta}")

        if resposta == "acertou":
            break
        elif resposta == "maior":
            baixo = chute + 1
        else:
            alto = chute - 1

    if mostrar:
        print(f"Acertei o {segredo} em {tentativas} tentativas. {avaliar(tentativas)}")
    return tentativas


random.seed(42)
jogar_robo()
````

Ele usa **as mesmas funções** `sortear`, `comparar` e `avaliar` que o jogo humano.
Nada foi reescrito.

Agora mil partidas, para ver se ele é bom mesmo:

````python
random.seed(7)
total = 0
pior = 0

for _ in range(1000):
    t = jogar_robo(mostrar=False)
    total += t
    if t > pior:
        pior = t

print(f"média: {total / 1000:.2f} tentativas")
print(f"pior caso: {pior}")
````

**Nunca mais que 7.** Nunca.

O motivo é bonito: cada chute no meio **corta o que sobrou pela metade**.

```
100 → 50 → 25 → 13 → 7 → 4 → 2 → 1
```

Sete cortes. E `2**7 = 128`, que é mais que 100.

Isso tem nome — **busca binária** — e é uma das ideias mais usadas em computação. Você acabou
de escrever uma sem saber.

# 💊 Pílula de IA — um modelo é uma função

## Etapa 1 — a caixa

Uma função é uma caixa: entra uma coisa, sai outra.

````python
def dobro(x):
    return x * 2


print(dobro(21))
````

Entra 21, sai 42. Você não precisa saber o que tem dentro para usar.

## Etapa 2 — o modelo da Aula 4 é literalmente isto

Lembra do laço que ajustava o peso até o erro cair? Ele parou em **1.681**.

O modelo treinado é esta função:

````python
def prever_nota(horas):
    return 1.681 * horas


print(prever_nota(2))
print(prever_nota(3.5))
````

É isso. **É o modelo inteiro.** Uma linha.

## Etapa 3 — a diferença está em quem escreveu o corpo

Coloque as duas lado a lado:

```python
def dobro(x):
    return x * 2          # ← o 2 fui EU que escrevi

def prever_nota(horas):
    return 1.681 * horas  # ← o 1.681 ninguém escreveu
```

O `1.681` saiu de **50 voltas de um `for`** medindo erro. Nenhum humano digitou aquele número.

Veja o que o treino fez, na prática — a chamada é idêntica, só o corpo mudou:

````python
def prever_chute(horas):
    return 0.5 * horas        # o chute inicial, antes de treinar


horas_reais = [2, 4, 6]
notas_reais = [4, 7, 10]

for i in range(len(horas_reais)):
    h = horas_reais[i]
    print(f"{h}h -> chute {prever_chute(h):.1f} | treinado {prever_nota(h):.1f} | real {notas_reais[i]}")
````

**Treinar um modelo é escrever o corpo de uma função sem digitar o corpo dela.**

Você define o formato (`peso * horas`) e deixa os dados escolherem os números.

## Etapa 4 — e é honesto dizer que ele erra

Peça a nota de quem estudou 7 horas:

````python
print(prever_nota(7))
````

**11.77.** Nota acima de 10.

O modelo não sabe que nota máxima é 10 — ninguém contou isso para ele. Ele só conhece a reta
que aprendeu, e os dados dele iam até 6 horas.

Isso se chama **extrapolar**, e é uma das formas mais comuns de IA falar bobagem com confiança
total. Toda vez que você pedir a um modelo algo fora do que ele viu, guarde essa desconfiança.

## Etapa 5 — a escala

Tudo que existe de IA cabe nessa forma:

| | Entra | Sai |
|---|---|---|
| `prever_nota` | horas | nota |
| detector de spam da Aula 3 | mensagem | spam / normal |
| reconhecedor de imagem | foto | "gato" |
| **ChatGPT** | os tokens da Aula 5 | o próximo token |

O ChatGPT é `def responder(tokens): return proximo_token`. A diferença não é a forma — é que o
corpo dele tem **bilhões** de números, e nenhum deles foi digitado por uma pessoa.

---

### O que fica da pílula de hoje

1. **Um modelo é uma função.** Entra dado, sai resposta.
2. **A sua função você escreveu; a do modelo os dados escreveram.** Só isso muda.
3. **Fora do que ele viu, ele inventa** — com a mesma cara de certeza. Um modelo não sabe que
   não sabe.
